```python
import pathlib as path
import pandas as pd
import easygui
```

```python
def filter_and_save_excel(raw_data_files, parameters):
    """Filters selected columns and saves them as separate sheets in a single Excel workbook."""

    if not raw_data_files or not parameters:                               # (1) Ensures data and parameters are available.
        print("Missing data or parameters. Aborting save.")
        return None

    output_file_path = easygui.filesavebox(                                # (2) Prompts the user to choose the output Excel file.
        msg="Choose where to save your filtered Excel file:",
        title="Save Filtered Data",
        default="Filtered_Data.xlsx",
        filetypes=["*.xlsx"]
    )

    if not output_file_path:                                               # (3) Stops if the user cancels the save dialog.
        print("Save operation cancelled.")
        return None

    output_file_path = path.Path(output_file_path).with_suffix(".xlsx")    # (4) Converts the selected path to a Path object
                                                                           #     and guarantees the .xlsx extension.

    print("=" * 90)                                                        # (5) Prints a separator line.
    print("Filtering and Saving Excel Workbook")                           # (6) Displays the current operation.
    print("=" * 90)                                                        # (7) Prints another separator line.

    with pd.ExcelWriter(output_file_path, engine="openpyxl") as writer:    # (8) Creates the output Excel workbook.

        for file_name, dataframe in raw_data_files.items():                # (9) Iterates through every loaded DataFrame.

            filtered_dataframe = dataframe.filter(items=parameters)         # (10) Filters the selected columns that exist in the current DataFrame.

            sheet_name = path.Path(file_name).stem[:31]                    # (11) Uses the original filename as the worksheet name.

            filtered_dataframe.to_excel(                                   # (12) Writes the filtered DataFrame to its worksheet.
                writer,
                sheet_name=sheet_name,
                index=False
            )

            print(f"✓ Added Sheet : {sheet_name}")                         # (13) Confirms the worksheet was created.
            print(f"  Columns     : {len(filtered_dataframe.columns)}")    # (14) Displays the number of retained columns.

    print("=" * 90)                                                        # (15) Prints a separator line.
    print(f"Workbook Saved : {output_file_path.name}")                     # (16) Displays the workbook name.
    print(f"Location       : {output_file_path.parent}")                   # (17) Displays the save folder.
    print("=" * 90)                                                        # (18) Prints the closing separator line.

    return output_file_path                                                # (19) Returns the full path of the saved workbook.
```
