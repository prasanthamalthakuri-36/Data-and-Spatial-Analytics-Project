```python
import easygui
```

```python
def choose_parameters(all_files):
    """Allows the user to select one or more column headings."""

    if not all_files:                                          # (1) Ensures data is available.
        print("No data available for parameter selection.")
        return None

    parameters = set()                                         # (2) Stores unique column headings.

    for dataframe in all_files.values():                       # (3) Iterates through every DataFrame.
        parameters.update(dataframe.columns)                   # (4) Adds its column headings to the set.

    parameters = sorted(parameters)                            # (5) Sorts the headings alphabetically.

    if not parameters:                                         # (6) Ensures at least one parameter exists.
        print("No parameters available for selection.")
        return None

    selections = easygui.multchoicebox(                        # (7) Displays a multi-selection dialog.
        title="PARAMETER SELECTION",
        msg="Select one or more parameters.",
        choices=parameters,
        preselect=None                                         # (8) Prevents any item from being selected initially.
    )

    if not selections:                                         # (9) Handles Cancel, window close, or no selection.
        print("Parameter selection cancelled.")
        return None

    print("=" * 90)                                            # (10) Prints a separator line.
    print("Selected Parameters")                               # (11) Displays the heading.
    print("=" * 90)                                            # (12) Prints another separator line.

    for number, parameter in enumerate(selections, start=1):   # (13) Iterates through the selected parameters with numbering.
        print(f"{number}. {parameter}")                        # (14) Prints each selected parameter.

    print("=" * 90)                                            # (15) Prints the closing separator line.

    return selections                                          # (16) Returns the selected column headings.
```
