```python
import pandas as pd
import pathlib as path
```

```python
def read_files(folder_path):
    """Reads all supported data files from the specified folder."""

    if not folder_path:                              # (1) Stops if folder selection was cancelled.
        return None

    readers = {                                     # (2) Maps each supported file extension to its Pandas reader.
        ".csv": pd.read_csv,
        ".json": pd.read_json,
        ".xml": pd.read_xml,
        ".xlsx": pd.read_excel
    }

    raw_files = {}                                  # (3) Stores each successfully loaded DataFrame.

    print("=" * 90)
    print("Reading Raw Data Files")
    print("=" * 90)

    for file_path in folder_path.iterdir():         # (4) Iterates through every file in the selected folder.

        try:
            reader = readers[file_path.suffix.lower()]      # (5) Selects the appropriate reader.
            raw_files[file_path.name] = reader(file_path)   # (6) Reads and stores the file.

            print(f"✓ Loaded : {file_path.name}")

        except Exception as error:
            print(f"✗ Failed : {file_path.name}")
            print(f"  Error  : {error}")

    print("=" * 90)
    print(f"Successfully Read : {len(raw_files)} file(s)")
    print("=" * 90)

    return raw_files                                  # (7) Returns all successfully loaded DataFrames.
```
