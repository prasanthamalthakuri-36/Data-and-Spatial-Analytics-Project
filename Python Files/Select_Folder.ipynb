```python
import easygui                # library to take user input
from pathlib import Path      # library to store folder path
```

```python
def select_folder():
    """
    Opens a folder selection dialog and returns
    the selected folder as a Path object.
    """

    folder = easygui.diropenbox(                    # (1) Opens the folder selection dialog box.
        title="FOLDER SELECTION",
        msg="Select the Raw Data Files Folder"
    )

    if folder:                                      # (2) Checks whether the user selected a folder.
        folder = Path(folder)                       # (3) Converts the selected folder path into a Path object.

        print("=" * 90)
        print("✓ Selected Folder Path")
        print(folder)
        print("=" * 90)

        return folder                               # (4) Returns the selected folder as a Path object.

    print("Folder selection cancelled.")            # (5) Displays a message if the selection is cancelled.
    return None                                     # (6) Returns None when no folder is selected.
```
