```python
import easygui
"""
from Select_Folder import select_folder
from Read_Files import read_files
from Choose_Parameters import choose_parameters
from Filter_Headers import filter_and_save_excel
from Text_Normalization import text_normalization
"""
```

```python
while True:

    try:

        # ==========================================================
        # MODULE 1 : SELECT RAW DATA FOLDER
        # ==========================================================

        raw_files_folder_path = select_folder()

        if not raw_files_folder_path:
            break


        # ==========================================================
        # MODULE 2 : READ RAW DATA FILES
        # ==========================================================

        raw_data_files = read_files(raw_files_folder_path)

        if not raw_data_files:
            break


        # ==========================================================
        # MODULE 3 : CHOOSE PARAMETERS
        # ==========================================================

        parameters = choose_parameters(raw_data_files)

        if not parameters:
            break


        # ==========================================================
        # MODULE 4 : FILTER PARAMETERS
        # ==========================================================

        filtered_excel_file_path = filter_and_save_excel(
            raw_data_files,
            parameters
        )

        if not filtered_excel_file_path:
            break


        # ==========================================================
        # MODULE 5 : TEXT NORMALIZATION
        # ==========================================================

        normalization = easygui.buttonbox(
            msg="Is text normalization required?",
            title="TEXT NORMALIZATION",
            choices=["Yes", "No"]
        )

        if normalization == "Yes":
            text_normalization(filtered_excel_file_path)


    except Exception as error:

        print("=" * 90)
        print("UNEXPECTED ERROR")
        print("=" * 90)
        print(error)
        print("=" * 90)


    if easygui.ynbox(
        msg="Do you want to analyse another data set?",
        title="SELECT OPTION"
    ):

        print("PREPARING THE NEXT DATASET.....")

    else:

        print("DATA ANALYSIS COMPLETE!")
        break
```
